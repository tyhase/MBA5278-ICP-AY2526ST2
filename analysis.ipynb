```python
from pathlib import Path
import pandas as pd


# Find CSV files contained in the repository
csv_files = [
    path for path in Path(".").rglob("*.csv")
    if ".git" not in path.parts
]

if len(csv_files) == 0:
    raise FileNotFoundError("No CSV dataset was found in the repository.")

if len(csv_files) > 1:
    raise ValueError(f"Multiple CSV files were found: {csv_files}")

data_path = csv_files[0]
print(f"Reading dataset: {data_path}")

# Read the provided dataset
df = pd.read_csv(data_path)

# Display the first five rows
display(df.head())

# Display summary statistics
display(df.describe(include="all"))
```
